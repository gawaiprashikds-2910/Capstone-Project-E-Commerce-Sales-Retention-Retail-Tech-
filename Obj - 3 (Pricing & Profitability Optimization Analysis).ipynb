# Objective - 3

### Task - 1 :- Calculate item-level profit margin after applying discounts and shipping costs across product categories and brands.

```python
profit_df = pd.merge(Order_item, Product, on="product_id", how="inner")
print(profit_df.head())
```

```python
profit_df["discount_amount"] = profit_df["unit_price"] * (profit_df["discount(%)"] / 100)

profit_df["final_selling_price"] = profit_df["unit_price"] - profit_df["discount_amount"]

profit_df.head()
```

```python
profit_df["profit"] = (
    profit_df["final_selling_price"]
    - profit_df["cost_price"]
    - profit_df["shipping_cost"]
)

profit_df.head()
```

```python
profit_df["profit_margin"] = (profit_df["profit"] / 
                              profit_df["final_selling_price"]) * 100

profit_df.head()
```

```python
profit_df["profit_margin"] = profit_df["profit_margin"].round(2)
profit_df.head()
```

```python
profit_df[
    [
        "product_id", "Category_name",
        "brand", "unit_price",
        "discount(%)", "shipping_cost",
        "cost_price", "final_selling_price",
        "profit", "profit_margin"
    ]
].head()
```

### Task 2 :- Analyze the impact of different discount bands on purchase conversion and average order value (AOV).

```python
df = Orders.merge(Order_item, on='order_id', how='inner')
df.head()
```

```python
df['discount_band'] = pd.cut(
    df['discount(%)'],
    bins=[0, 10, 20, 30, 100],
    labels=['0-10%', '11-20%', '21-30%', 'Above 30%'],
    include_lowest=True
)

df.head()
```

```python
df['conversion'] = df['order_status'].apply(
    lambda x: 1 if x == 'delivered' else 0
)

df.head()
```

```python
df['order_value'] = (df['unit_price'] - (df['unit_price'] * df['discount(%)'] / 100)
                     + df['shipping_cost'])

df.head()
```

```python
result = df.groupby('discount_band').agg(Purchase_Conversion_Rate=('conversion', 'mean'),
                                         Average_Order_Value=('order_value', 'mean'),
                                         Total_Orders=('order_id', 'count')).reset_index()

result.head()
```

```python
result['Purchase_Conversion_Rate'] = (result['Purchase_Conversion_Rate'] * 100).round(2)

result['Average_Order_Value'] = result['Average_Order_Value'].round(2)

result
```

### Task 3 :- Evaluate how shipping costs influence customer purchases and overall profitability.

```python
df1 = pd.merge(Order_item, Product, on='product_id', how='inner')
df1.head()
```

```python
df1['final_price'] = df1['unit_price'] - (df1['unit_price'] * df1['discount(%)'] / 100)

df1['profit'] = df1['final_price'] - df1['cost_price'] - df1['shipping_cost']

df1.head()
```

```python
df1['shipping_band'] = pd.cut(df1['shipping_cost'],
                              bins=[0, 50, 100, 200, float('inf')],
                              labels=['Low', 'Medium', 'High', 'Very High'],
                              include_lowest=True)

df1.head()
```

```python
result = df1.groupby('shipping_band').agg(Total_Orders=('order_id', 'count'),
                                          Average_Shipping_Cost=('shipping_cost', 'mean'),
                                          Average_Order_Value=('final_price', 'mean'),
                                          Total_Profit=('profit', 'sum'),
                                          Average_Profit=('profit', 'mean')).reset_index()

result.head()
```

```python
result[['Average_Shipping_Cost', 'Average_Order_Value',
        'Total_Profit', 'Average_Profit']] = result[['Average_Shipping_Cost', 'Average_Order_Value',
                                                     'Total_Profit', 'Average_Profit']].round(2)

result
```

### Task 4 :-  Identify the optimal pricing and discount strategy for maximizing revenue and profit across customer segments.

```python
Customer = pd.read_excel("Customers_dataset.xlsx")
Customer.head()
```

```python
df2 = Orders.merge(Customer, on='customer_id', how='inner')
df2.head()
```

```python
df2 = df2.merge(Order_item, on='order_id', how='inner')
df2 = df2.merge(Product, on='product_id', how='inner')

df2.head()
```

```python
df2['discount_amount'] = df2['unit_price'] * (df2['discount(%)'] / 100)
df2.head()
```

```python
df2['final_price'] = df2['unit_price'] - df2['discount_amount']

df2['revenue'] = df2['final_price'] * df2['quantity']

df2.head()
```

```python
df2['profit'] = ((df2['final_price'] - df2['cost_price'] - df2['shipping_cost'])
                * df2['quantity'])

df2.head()
```

```python
df2['discount_band'] = pd.cut(df2['discount(%)'],
                       bins=[0, 10, 20, 30, 100],
                       labels=['0-10%', '11-20%', '21-30%', 'Above 30%'],
                       include_lowest=True)

df2.head()
```

```python
result = df2.groupby(['customer_segment', 'discount_band']).agg(Total_Revenue=('revenue', 'sum'),
                                                                Total_Profit=('profit', 'sum'),
                                                                Average_Selling_Price=('final_price', 'mean'),
                                                                Total_Orders=('order_id', 'count')
                                                                ).reset_index()

result.head()
```

```python
result['Total_Revenue'] = result['Total_Revenue'].round(2)
result['Total_Profit'] = result['Total_Profit'].round(2)
result['Average_Selling_Price'] = result['Average_Selling_Price'].round(2)

result
```
